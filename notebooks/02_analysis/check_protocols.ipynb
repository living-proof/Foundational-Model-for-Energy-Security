```python
import subprocess
from collections import Counter
from pathlib import Path
import os
```

```python
clean_data = "data\\raw\\Modbus Dataset\\benign\\ied1a\\ied1a-network-capture\\veth4edc015-normal-4.pcap"
mitm_data = "data\\raw\\Modbus Dataset\\attack\\external\\ied1a\\ied1a-network-capture\\veth4edc015-0.pcap"

absolute_clean_path = (Path("../..") / clean_data).resolve()
absolute_mitm_path = (Path("../..") / mitm_data).resolve()

# Print the absolute paths
print("Absolute path to clean data:", absolute_clean_path)
print("Absolute path to mitm data:", absolute_mitm_path)
```

```python
def get_protocols(pcap_file):
    """Get protocol counts from a pcap file"""
    result = subprocess.run(
        ['tshark', '-r', pcap_file, '-T', 'fields', '-e', '_ws.col.Protocol'],
        capture_output=True,
        text=True
    )
    
    protocols = result.stdout.strip().split('\n')
    counts = Counter(protocols)
    return counts
```

```python
# Usage
pcap_file = absolute_clean_path
protocols = get_protocols(pcap_file)
print(protocols)

pcap_file = absolute_mitm_path
protocols = get_protocols(pcap_file)
print(protocols)
```
