```python
import pandas as pd
from datetime import datetime
from pathlib import Path
import csv
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Attack log paths
# Define relative paths
relative_attack_log_path = "data\\raw\\Modbus Dataset\\attack\\compromised-scada\\attack logs\\03-15-2023\\03-15-2023-1.csv"
relative_attack_log_path_2 = "data\\raw\\Modbus Dataset\\attack\\compromised-scada\\attack logs\\03-16-2023\\03-16-2023-1.csv"

absolute_attack_log_path = (Path("../../..") / relative_attack_log_path).resolve()
absolute_attack_path_2 = (Path("../../..") / relative_attack_log_path_2).resolve()

# Print the absolute paths
print("Absolute path to attack log data:", absolute_attack_log_path)
print("Absolute path to second attack log data:", absolute_attack_path_2)
```

```python
# Processed Modbus Compromised SCADA paths
relative_modbus_compromised_scada_path = "data\\processed\\Modbus Dataset\\attack\\compromised-scada\\ied1a\\MODBUS_ONLY_veth3614724-0.csv"

absolute_modbus_compromised_scada_path = (Path("../../..") / relative_modbus_compromised_scada_path).resolve()

print("Absolute path to processed Modbus Compromised SCADA data:", absolute_modbus_compromised_scada_path)
```

```python
# Load attack log data into DataFrames
attack_log_03_15_2023_df = pd.read_csv(absolute_attack_log_path)
attack_log_03_16_2023_df = pd.read_csv(absolute_attack_path_2)

print("Attack log 03-15-2023 DataFrame shape:", attack_log_03_15_2023_df.shape)
print("Attack log 03-16-2023 DataFrame shape:", attack_log_03_16_2023_df.shape)

# Combine the two attack log DataFrames
combined_attack_log_df = pd.concat([attack_log_03_15_2023_df, attack_log_03_16_2023_df], ignore_index=True)
print("Combined attack log DataFrame shape:", combined_attack_log_df.shape)

# Filter the combined attack log DataFrame for entries with TargetIP and convert Timestamp to datetime
filtered_combined_attack_log_df = combined_attack_log_df[combined_attack_log_df['TargetIP'] == '185.175.0.4'].copy()
filtered_combined_attack_log_df['datetime'] = pd.to_datetime(filtered_combined_attack_log_df['Timestamp'], utc=True)
print("Filtered combined attack log DataFrame shape:", filtered_combined_attack_log_df.shape)

display(filtered_combined_attack_log_df.head(10))
```

```python
# Load the processed Modbus Compromised SCADA DataFrame
modbus_comprimised_scada_df = pd.read_csv(absolute_modbus_compromised_scada_path)
print("Processed Modbus Compromised SCADA DataFrame shape:", modbus_comprimised_scada_df.shape)

# Filter the Modbus Compromised SCADA DataFrame for entries with dst_ip and src_ip
filtered_modbus_comprimised_scada_df = modbus_comprimised_scada_df[
    (modbus_comprimised_scada_df['dst_ip'] == '185.175.0.4') & 
    (modbus_comprimised_scada_df['src_ip'] == '185.175.0.3')
]

# Print the shape of the filtered Modbus Compromised SCADA DataFrame and display the first 10 rows
print("Filtered Modbus Compromised SCADA DataFrame shape:", filtered_modbus_comprimised_scada_df.shape)
display(filtered_modbus_comprimised_scada_df.head(10))
```

```python
# Ensure both datetime columns are the same type
modbus_comprimised_scada_df['datetime'] = pd.to_datetime(modbus_comprimised_scada_df['datetime'], utc=True)
filtered_combined_attack_log_df['datetime'] = pd.to_datetime(filtered_combined_attack_log_df['datetime'], utc=True)

# Filter by IP first (reduces data size)
ip_mask = (modbus_comprimised_scada_df['src_ip'] == '185.175.0.3') & \
          (modbus_comprimised_scada_df['dst_ip'] == '185.175.0.4')

# Use merge_asof for fast time-based matching
modbus_comprimised_scada_df = modbus_comprimised_scada_df.sort_values('datetime')
attack_times = filtered_combined_attack_log_df[['datetime']].sort_values('datetime')

# Find nearest attack time for each packet
merged = pd.merge_asof(
    modbus_comprimised_scada_df,
    attack_times.assign(attack_time=attack_times['datetime']),
    on='datetime',
    direction='nearest',
    tolerance=pd.Timedelta(milliseconds=100)
)

# Label based on whether a match was found AND IP matches
modbus_comprimised_scada_df['label'] = np.where(
    merged['attack_time'].notna() & ip_mask,
    'ANOMALY',
    'NORMAL'
)

# Print the shape of the Modbus Compromised SCADA DataFrame and display the first 10 rows
print("Filtered Modbus Compromised SCADA DataFrame shape:", modbus_comprimised_scada_df.shape)
display(modbus_comprimised_scada_df.head(10000))
```

```python
# Save the labeled Modbus Compromised SCADA DataFrame to a new CSV file
modbus_comprimised_scada_df.to_csv(absolute_modbus_compromised_scada_path, index=False)
```
